## Imports

```python
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
import plotly.io as pio
pio.renderers.default = "plotly_mimetype+notebook_connected"
```

## Auxiliary functions

```python
def plot_3d(meshes, width=700, height=600):

    colorscales=["Teal", "Peach", "Greens", "Magenta", "Oranges", "Purp"]

    fig = go.Figure()

    for i, mesh in enumerate(meshes):

        # add trace
        fig.add_trace(
            go.Surface(
                x=mesh[0],
                y=mesh[1],
                z=mesh[2],
                opacity=1.0,
                showscale=False,
                reversescale=True,
                lighting=dict(diffuse=0.5, roughness=0.5, ambient=0.65, specular=0),
                colorscale=colorscales[i],
                hovertemplate="(%{x:.2f}, %{y:.2f}, %{z:.2f})<extra></extra>",
                contours=dict(
                    x=dict(
                        show=False,
                        highlight=False
                    ),
                    y=dict(
                        show=False,
                        highlight=False
                    ),
                    z=dict(
                        show=False,
                        highlight=False
                    )
                )
            )
        )

    # format axes for beauty
    # this applies here https://plotly.com/python/configuration-options/
    fig.update_layout(

        margin=go.layout.Margin(
            l=0, #left margin
            r=0, #right margin
            b=0, #bottom margin
            t=0, #top margin
        ),

        # we do not use hard width and height values because
        # this prevents the plot for being responsive
        #width=something,
        #height=something,

        hoverlabel_align="right",

        scene=dict(
            xaxis=dict(
                title="x",
                backgroundcolor="rgb(255, 255, 255)",
                gridcolor="rgb(170,170,170)",
                zerolinecolor="rgb(255,255,255)",
                showspikes=False,
                showticklabels=False,
            ),
            yaxis=dict(
                title="y",
                backgroundcolor="rgb(255, 255, 255)",
                gridcolor="rgb(170,170,170)",
                zerolinecolor="rgb(255,255,255)",
                showspikes=False,
                showticklabels=False,
            ),
            zaxis=dict(
                title="z",
                backgroundcolor="rgb(255, 255, 255)",
                gridcolor="rgb(170,170,170)",
                zerolinecolor="rgb(255,255,255)",
                showspikes=False,
                showticklabels=False,
            )
        )

    )

    return fig
```

## Main body

$$
\begin{align}
\text{Blue Waves} &: -\frac{\sin(x^2+y^2)}{x^2+y^2} \\
\text{Orange Sphere} &: x^2 + y^2 + (z-1)^2 - (0.5)^2 = 0 \\
\text{Green Sphere} &: (x-0.5)^2 + (y-0.5)^2 + (z-0.5)^2 - (0.3)^2 = 0 \\
\text{Purple Sphere} &: (x-0.5)^2 + (y+0.5)^2 + (z-0.25)^2 - (0.2)^2 = 0
\end{align}
$$

```python
# Blue Waves
x1_axis = np.arange(-3, 3, 0.1)
y1_axis = np.arange(-3, 3, 0.1)
x1_mesh, y1_mesh = np.meshgrid(x1_axis, y1_axis)
z1_mesh = -np.sin(x1_mesh**2+y1_mesh**2)/(x1_mesh**2+y1_mesh**2)

# Orange Sphere
theta_axis = np.linspace(0, 2*np.pi, 100)
phi_axis = np.linspace(0, np.pi, 100)
theta_mesh, phi_mesh = np.meshgrid(theta_axis, phi_axis)
x2_mesh = 0.5*np.cos(theta_mesh)*np.sin(phi_mesh)
y2_mesh = 0.5*np.sin(theta_mesh)*np.sin(phi_mesh)
z2_mesh = 0.5*np.cos(phi_mesh) + 1

# Green Sphere
x3_mesh = 0.3*np.cos(theta_mesh)*np.sin(phi_mesh) + 0.5
y3_mesh = 0.3*np.sin(theta_mesh)*np.sin(phi_mesh) + 0.5
z3_mesh = 0.3*np.cos(phi_mesh) + 0.5

# Purple Sphere
x4_mesh = 0.2*np.cos(theta_mesh)*np.sin(phi_mesh) + 0.5
y4_mesh = 0.2*np.sin(theta_mesh)*np.sin(phi_mesh) - 0.5
z4_mesh = 0.2*np.cos(phi_mesh) + 0.25

# plot meshes
fig = plot_3d([
    [x1_mesh, y1_mesh, z1_mesh],
    [x2_mesh, y2_mesh, z2_mesh],
    [x3_mesh, y3_mesh, z3_mesh],
    [x4_mesh, y4_mesh, z4_mesh]
])

fig.show()
```
